## Stage 5.Univariate Analysis

📌 Definition

Univariate analysis refers to the analysis of a single variable (one feature) to understand its characteristics and distribution.

🎯 Purpose

The main goal is to summarize, describe data and Understand the distribution of values.

📊 Key Techniques in Univariate Analysis
1. Summary Statistics that is used to describe the data numerically.
2. Frequency Distribution that shows how often each value occurs.
3. Visualization helps understand data visually.

🧠 Quick Summary
Focus: one variable
Goal: describe and understand data
Methods: statistics + charts


'******************************************************************************'



#### 👉Types of Univariate Analysis:
---


1️⃣ Numerical (Quantitative) Analysis: represents measurable values (numbers) like age, salary, height, etc. Supports mathematical operations (add, subtract, average)
Types:
  Discrete → countable (e.g., number of users)
  Continuous → measurable (e.g., temperature)

2️⃣ Categorical (Qualitative) Analysis: represents labels or groups (non-numeric values) like gender, region, product type. Cannot perform arithmetic operations
  Types:
  Nominal → no order (e.g., color)
  Ordinal → ordered categories (e.g., rating: low/medium/high)





### 5A.Numerical Analysis : Statistics & Analysis

```python
#Summary statistics for numeric columns
#numeric_cols=['employee_count','annual_revenue_million','profit_margin_percent','customer_count','market_rating']

#Numeric cols with excluding 2 numeric cols
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
exclude=["id","founded_year"]
numeric_cols = [col for col in numeric_cols if col not in exclude]

print("**************** Summary statistics for numeric columns ****************")

summary=df[numeric_cols].describe().T
summary['mean']=df[numeric_cols].mean()
summary['median']=df[numeric_cols].median()
summary['mode']=df[numeric_cols].mode().iloc[0]
summary['min']=df[numeric_cols].min()
summary['max']=df[numeric_cols].max()
summary['std']=df[numeric_cols].std()
summary['var']=df[numeric_cols].var()
summary
```

```python
#Histogram, boxplot for numerical data
for col in numeric_cols:
  plt.figure(figsize=(8,4))

  plt.subplot(1,2,1)
  sns.histplot(df[col],kde=True,color="#A9A9A9")
  plt.title(f'Histogram of {col}')

  plt.subplot(1,2,2)
  sns.boxplot(df[col])
  plt.title(f'Boxplot of {col}')

  plt.tight_layout()
  plt.show()

  print(f"\nInsight for {col}:")
  print("👉 Data appears symmetric with values centered around the middle")
  print("👉 No strong skewness or extreme outliers observed that indicates data is approximately normally distributed.\n\n")
```

### 5C. Categorical Analysis

```python
# Horizontal BarChart for showing categorical data
categorical_cols = ['industry', 'country', 'region']

# Create subplots (1 row, 3 columns)
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

for i, col in enumerate(categorical_cols):
    if col != 'region':
        sns.countplot(
            y=col,
            data=df,
            order=df[col].value_counts().index,
            ax=axes[i],
            color='lightcoral'
        )
        axes[i].set_title(f'Bar Chart of {col}')
        axes[i].tick_params(axis='x', rotation=45)

    else:
        # Donut chart for region
        counts = df[col].value_counts()
        axes[i].pie(
            counts,
            labels=counts.index,
            autopct="%1.1f%%",
            colors=plt.cm.Set3.colors,
            wedgeprops={"width": 0.5, "edgecolor": "white"}
        )
        axes[i].set_title(f"Donut Chart of {col}")

plt.tight_layout()
plt.show()

# Separate insights
for col in categorical_cols:
    print(f"\nInsight for {col}:")

    if col != 'region':
        top_category = df[col].value_counts().idxmax()
        top_count = df[col].value_counts().max()
        total = df[col].value_counts().sum()
        percentage = (top_count / total) * 100

        print(f"👉 The bar chart shows the distribution of '{col}' categories and '{top_category}' is the most frequent category with {top_count} records ({percentage:.1f}%).")

    else:
        top_category = df[col].value_counts().idxmax()
        top_count = df[col].value_counts().max()
        total = df[col].value_counts().sum()
        percentage = (top_count / total) * 100

        print(f"👉 The donut chart shows the percentage share of each region and '{top_category}' has the highest contribution with {percentage:.1f}% of the total data.")
```

```python
#Categorical data in Tabular Format
for i, col in enumerate(categorical_cols):
    # Get frequency counts
    freq_table = df[col].value_counts().reset_index()
    freq_table.columns = [col, 'Count']

    print(f"\nFrequency count for {col}:")
    print(freq_table.to_string(index=False))
```

### 6A. Numerical vs Numerical

```python
#Numeric cols vs Numeric cols
# Number of pairs
pairs = [(numeric_cols[i], numeric_cols[j]) for i in range(len(numeric_cols)) for j in range(i+1, len(numeric_cols))]

# Define grid size
n = len(pairs)
cols = 3   # number of plots per row
rows = (n + cols - 1) // cols

fig, axes = plt.subplots(rows, cols, figsize=(15, 5*rows))
axes = axes.flatten()

# Loop through pairs and plot
for idx, (col_x, col_y) in enumerate(pairs):
    ax = axes[idx]
    sns.regplot(
        data=df,
        x=col_x,
        y=col_y,
        scatter_kws={"alpha":0.6, "color":"steelblue"},
        line_kws={"color":"red"},
        ax=ax
    )
    ax.set_title(f"{col_x} vs {col_y} (Linear Fit)")
    ax.set_xlabel(col_x)
    ax.set_ylabel(col_y)

# Hide unused subplots
for j in range(idx+1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
print("\n👉 All the variables show a balanced distribution around the median\n")
print("👉 Indicating no strong skewness and equally distributed around the median\n")
```

```python
#Founded Year vs Numerical cols
# Ensure founded_year is numeric
df["founded_year"] = pd.to_numeric(df["founded_year"], errors="coerce")

# Group by founded_year and calculate mean for numeric columns
avg_df = df.groupby("founded_year").mean(numeric_only=True).reset_index()

print("Average values by Founded Year:")
print(avg_df.head())

# Prepare subplot grid
n = len(numeric_cols)
cols = 3   # number of plots per row
rows = (n + cols - 1) // cols

fig, axes = plt.subplots(rows, cols, figsize=(15, 5*rows))
axes = axes.flatten()

# Loop through numeric columns and plot
for idx, col in enumerate(numeric_cols):
    ax = axes[idx]
    sns.regplot(
        data=avg_df,
        x="founded_year",
        y=col,
        scatter_kws={"alpha":0.6, "color":"steelblue"},
        line_kws={"color":"red"},
        ax=ax
    )
    ax.set_title(f"Average {col} vs Founded Year")
    ax.set_xlabel("Founded Year")
    ax.set_ylabel(f"Average {col}")

# Hide unused subplots if any
for j in range(idx+1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()

print("\n\n✅ One-line insight for growing trend report")
print("👉Employee Count, Profit Margin, and Market Rating all show a positive relationship with Founded Year, indicating that these \nvariables increase together over time. This suggests a consistent growth pattern where companies tend to expand their workforce, \nimprove profitability, and achieve higher market ratings simultaneously, reflecting overall business maturity and performance improvement.")
print("\n\n✅ One-line insight for Falling trend report")
print("👉Annual revenue and customer count show a negative correlation with founded year, indicating that newer companies tend to have\nsmaller customer bases and lower revenue, while older companies have established market presence and higher revenue levels")
```

🔎 Key Characteristics

#### Correlation Heatmap
A Correlation Heatmap is a visual representation that shows the strength and direction of relationships between multiple variables using color-coded values.

Each cell represents the correlation coefficient between two variables.
Values range from -1 to +1:

+1 → strong positive relationship

-1 → strong negative relationship

0 → no relationship


Colors help quickly identify patterns:

Warmer colors (e.g., red) indicate strong correlation

Cooler colors (e.g., blue) indicate weak or no correlation

```python
#Correlation Heatmap of Numeric cols vs Numeric cols
num_df=df[numeric_cols]
corr_matrix = num_df.corr(method='pearson')

plt.figure(figsize=(7, 5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5,fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()

print("\n👉The correlation heatmap indicates that all variables are essentially uncorrelated, with correlation coefficients close to zero.\nThis suggests there are no significant linear relationships between employee count, revenue, profit margin, customer count, and market rating, and \neach variable behaves independently within the dataset")
```

##🔍 Stage 8. Outlier Detection
Outliers in the dataset were identified using three standard techniques: IQR, Z-score, and Boxplots. The combined use of these methods provides a robust and reliable detection framework

• Interquartile Range (IQR) Method  
• Z-score Method  
• Boxplots   

###📊 8A. Interquartile Range (IQR) Method

Detects outliers based on data spread using:

Lower bound = Q1 − 1.5 × IQR
Upper bound = Q3 + 1.5 × IQR


✅ Observations:

Effectively identifies extreme values outside the central 50% of data.
Particularly useful for skewed distributions.

```python
# Loop through numeric columns
for col in numeric_cols :
    print(f"\n<-------📊 {col} 📊------->")
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    IQR = q3 - q1
    lower_bound = q1 - 1.5 * IQR
    upper_bound = q3 + 1.5 * IQR
    print("q1: ",q1)
    print("q3: ",q3)
    print("IQR: ",IQR)
    print("lower_bound: ",lower_bound)
    print("upper_bound: ",upper_bound)

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: {len(outliers)} outliers")
    print("="*50)

print("\n👉 The result shows: 0 outliers detected using the IQR method")

print("\n✅ Primary Insight: The dataset is clean, consistent, and free from extreme values based on statistical thresholds.")
print("\nSince rating scale is bounded likely 0–5, all values fall within range → no outliers detected")
```

###📈 8B. Z‑Score Method

Measures how far a data point is from the mean (in standard deviations).
Common threshold: |Z-score| > 3 → outlier

✅ Insight:

Suitable for normally distributed data.
Helps identify global outliers that significantly deviate from the average.

```python
# Function to detect outliers using Z-score
def detect_outliers_zscore(df):
    outliers = {}

    for col in numeric_cols:
        mean = df[col].mean()
        std = df[col].std()
        z_scores = (df[col] - mean) / std
        print(f"\n<-------📈 {col} 📈------->")
        print("mean:",mean)
        print("std:",std)
        print("Min of z_score: ",min(z_scores))
        print("Max of z_score: ",max(z_scores))
        print("="*50)
        threshold = 3
        outliers[col] = df[np.abs(z_scores) > threshold][col].tolist()

    return outliers

# Detect outliers
outliers = detect_outliers_zscore(df)
print("Outliers detected:", outliers)

print("\n✅The Z-score analysis confirms that all variables fall within ±2 standard deviations, indicating a well-balanced and normally distributed dataset with no outliers.")
print("\n✅Combined with IQR results, this strongly suggests high data consistency and quality, but also indicates limited variability and absence of extreme cases")
```

##Stage 9.Skewness & Kurtosis

🔍 1. Skewness Insight (Distribution Shape)

✅ What Skewness Indicates:

Skewness ≈ 0 → symmetric distribution

Positive skew (>0) → long right tail

Negative skew (<0) → long left tail



📈 2. Kurtosis Insight (Tail Behavior / Peaks)

✅ What Kurtosis Indicates:

Kurtosis ≈ 3 (Mesokurtic) → normal distribution

Kurtosis > 3 (Leptokurtic) → heavy tails (more outliers)

Kurtosis < 3 (Platykurtic) → light tails (fewer outliers)

```python
#In Tabular Format
results = []

for col in numeric_cols:
    results.append({
        "Column": col,
        "Skewness": df[col].skew(),
        "Kurtosis": df[col].kurtosis()
    })

skew_kurt_df = pd.DataFrame(results)
print(skew_kurt_df)
```

```python
# In Visual Format
stats = []
for col in numeric_cols:
    stats.append({
        "Column": col,
        "Skewness": df[col].skew(),
        "Kurtosis": df[col].kurtosis()
    })

stats_df = pd.DataFrame(stats)

# Melt the dataframe for grouped bar plotting
melted = stats_df.melt(id_vars="Column", value_vars=["Skewness","Kurtosis"],
                       var_name="Statistic", value_name="Value")

plt.figure(figsize=(12,6))
ax = sns.barplot(data=melted, x="Column", y="Value", hue="Statistic", palette="Set2")

plt.title("Skewness & Kurtosis of Numeric Columns")
plt.xticks(rotation=45)

# Add value labels on each bar
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f")

plt.show()

print("\n✅ Skewness Insight Interpretation:\n1.A skewness value near 0 means the distribution is symmetric\n2.There is no meaningful right-skew or left-skew in any variable\n3.Positive and negative values are both very small, so none of the variables are noticeably imbalanced")
print("\n✅ Kurtosis Insight Interpretation:The chart shows that all variables are nearly perfectly symmetric (skewness close to 0) and platykurtic (kurtosis \naround -1.2), meaning the dataset is flat, evenly distributed, and free from extreme values. This confirms your earlier findings from the IQR, Z-score,\nand boxplot analyses that the data is statistically clean, stable, and well-suited for modeling or descriptive analysis.")
```
